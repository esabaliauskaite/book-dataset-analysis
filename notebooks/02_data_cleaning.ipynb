## 1. Cleaning Decision Table

| Column / area      | Raw problem                                                      | Cleaning decision                                  | Output / result                                  | Priority     |
| ------------------ | ---------------------------------------------------------------- | -------------------------------------------------- | ------------------------------------------------ | ------------ |
| Whole dataset      | Exact duplicate rows exist                                       | Remove exact duplicates only                       | Cleaner base dataset                             | High         |
| `bookId`           | Some duplicate IDs                                               | Investigate before dropping                        | Decide whether duplicate IDs are true duplicates | High         |
| `title + author`   | Many duplicates                                                  | Keep for now; likely editions/formats/translations | No automatic deletion                            | High         |
| `author`           | Main authors mixed with illustrators, translators, introductions | Separate primary authors from contributors         | New column `primary_authors`, `contributors`     | High         |
| `genres`           | Stored as stringified list                                       | Parse into real list                               | `genres_list`                                    | High         |
| `ratingsByStars`   | Stored as list-like string                                       | Convert into five numeric star-count columns       | `star_5`, `star_4`, `star_3`, `star_2`, `star_1` | High         |
| `publishDate`      | Messy formats and two-digit years                                | Extract year, flag parse quality                   | `publish_year`, `publish_date_quality`           | High         |
| `firstPublishDate` | Important but many missing values                                | Extract year, preserve missingness                 | `first_publish_year`                             | High         |
| `rating`           | Numeric but needs validity check                                 | Validate range 0–5                                 | Keep as numeric                                  | High         |
| `numRatings`       | Popularity measure, may have extreme skew                        | Validate non-negative, maybe log-transform later   | `numRatings`, later `log_numRatings`             | High         |
| `bbeScore`         | Ranking/popularity feature                                       | Validate and preserve                              | Keep raw, maybe normalize later                  | High         |
| `bbeVotes`         | May contain invalid negative values                              | Investigate negative values before using           | Cleaned or flagged value                         | High         |
| `description`      | Needed for NLP; may contain missing/short text                   | Audit length, language, duplicates, noise          | `description_clean`, `description_length`        | High for NLP |
| `isbn`             | Contain placeholders and maybe malformed values                  | Keep raw ISBN, create cleaned ISBN validity flag   | New column `isbn_valid`                          | Medium       |
| `awards`           | Stored as stringified list                                       | Parse; create award count and award flag           | `awards_list`, `award_count`, `has_award`        | Medium       |
| `pages`            | Numeric-like but may contain invalid values                      | Convert to numeric, flag bad values                | `pages_num`, `pages_valid`                       | Medium       |
| `likedPercent`     | Percentage field                                                 | Validate 0–100                                     | Keep or convert to proportion                    | Medium       |
| `language`         | Many unique values, possible inconsistency                       | Normalize categories; group rare languages         | `language_clean`                                 | Medium       |
| `bookFormat`       | Explains edition duplicates                                      | Standardize values                                 | `book_format_clean`                              | Medium       |
| `series`           | Missingness may mean standalone book                             | Create `is_series` flag                            | `is_series`                                      | Medium       |
| `price`            | High missingness and unclear meaning                             | Parse only if needed later                         | `price_parsed`                                   | Low/Medium   |
| `characters`       | Stored as stringified list;                                      | Parse only if needed later                         | `characters_list`                                | Low          |
| `setting`          | Stored as stringified list;                                      | Parse but do not prioritize for first EDA          | `setting_list`                                   | Low          |
| `edition`          | Very high missingness                                            | Parse only if needed later                         | `edition_parsed`                                 | Low          |
