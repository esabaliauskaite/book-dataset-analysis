## 7. Dataset Grain and Duplicate Strategy

Before cleaning or removing duplicate records, I need to define what one row represents. In this dataset, one row appears to represent a Goodreads book record or edition-level entry, not necessarily one unique literary work. This distinction matters because the same book title and author may appear multiple times due to different editions, formats, translations, publication dates, or ISBNs.

| Possible grain     | Meaning                                                | Consequence                                 |
| ------------------ | ------------------------------------------------------ | ------------------------------------------- |
| Record-level       | Each row is one scraped Goodreads record               | Safest assumption for now                   |
| Edition-level      | Each row is a specific edition or format of a book     | Same title/author can appear multiple times |
| Work-level         | Each row is one unique book/work regardless of edition | Would require merging editions              |
| Author-title-level | Each row is one title-author combination               | Too simplistic and risky                    |


I’ll be concentrating on handling the dataset as individual records or editions. Duplicate titles alone are not reliable indicators of duplicate books because different books can share the same or similar titles.

```python
def has_5_elements(value):
    try:
        parsed = ast.literal_eval(value)
        return isinstance(parsed, list) and len(parsed) == 5
    except (ValueError, SyntaxError, TypeError):
        return False
    
def has_0_elements(value):
    try:
        parsed = ast.literal_eval(value)
        return isinstance(parsed, list) and len(parsed) == 0
    except (ValueError, SyntaxError, TypeError):
        return False    
```

```python
def sum_list_string(value):
    try:
        parsed = ast.literal_eval(value)

        if not isinstance(parsed, list):
            return None

        return sum(float(x) for x in parsed)

    except (ValueError, SyntaxError, TypeError):
        return None
```

```python
ratings_count_check = df_raw['ratingsByStars'].apply(has_5_elements)
```

```python
null_ratings_count_check = df_raw['ratingsByStars'].apply(has_0_elements)
```

```python
sum_star_ratings = df_raw["ratingsByStars"].apply(sum_list_string)
matches_numRatings = sum_star_ratings == df_raw['numRatings']
```

```python
print("ratingsByStars with 5 elements:",ratings_count_check.sum())
print("ratingsByStars with 0 elements:",null_ratings_count_check.sum())
print("ratingsByStars matches numRatings:",matches_numRatings.sum())
```

### 10.5 Date validity

```python
missing = (publish_date.isna() | publish_date.astype(str).str.strip().eq("") | publish_date.astype(str).str.strip().eq("Unknown") ).sum()

year_only_pattern = r"^\d{4}$"
year_only = publish_date.astype(str).str.strip().str.fullmatch(year_only_pattern).sum()

month_year_text_pattern = r"^(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+\d{4}$"
month_year_text = publish_date.astype(str).str.strip().str.fullmatch(month_year_text_pattern).sum()

month_day_year_text_pattern = r"^(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+(?:(?:1|21|31)st|(?:2|22)nd|(?:3|23)rd|(?:[4-9]|1[0-9]|20|2[4-9]|30)th)(?:,\s*)?\s+\d{4}$"
month_day_year_text = publish_date.astype(str).str.strip().str.fullmatch(month_day_year_text_pattern).sum()

two_number_year_pattern = r"\b(?:0?[1-9]|1[0-2])/(?:0?[1-9]|[12][0-9]|3[01])/\d{2}\b"
numeric_mmddyy = publish_date.astype(str).str.strip().str.fullmatch(two_number_year_pattern).sum()

four_number_year_pattern = r"\b(?:0?[1-9]|1[0-2])/(?:0?[1-9]|[12][0-9]|3[01])/\d{4}\b"
numeric_mmddyyyy = publish_date.astype(str).str.strip().str.fullmatch(four_number_year_pattern).sum()

unknown_format = len(publish_date) - (missing + year_only + month_year_text + month_day_year_text + numeric_mmddyy + numeric_mmddyyyy)

print('Missing dates: ', missing)
print('Unknown format dates: ', unknown_format)
print('Year only dates: ', year_only)
print('Month-year text dates: ', month_year_text)
print('Month-day-year text dates: ', month_day_year_text)
print('Numeric MM/DD/YY dates ', numeric_mmddyy)
print('Numeric MM/DD/YYYY dates: ', numeric_mmddyyyy)
```

The remaining unknown `firstPublishDate` values are not ordinary formatting errors. They include one malformed partial date, two short historical year values, and one negative year that likely represents BCE. These should not be force-parsed with the main date patterns.

For the cleaning stage, these values should be handled separately from standard date formats. Some may still contain useful year information, but they require a special quality flag because they are not directly comparable to normal modern publication dates without interpretation.

```python
def parse_year_and_flag(date_series):
    clean = date_series.fillna("").astype(str).str.strip()
    placeholder_set = set(placeholder_values_lower)

    missing_mask = date_series.isna() | clean.eq("") | clean.str.lower().isin(placeholder_set) | clean.str.contains("Published", na=False) | clean.str.contains("Books", na=False) | clean.str.contains("books", na=False) | clean.str.contains("Expected publication:", na=False)
    full_date_month_mask = clean.str.fullmatch(month_day_year_text_pattern).fillna(False)
    full_date_numeric_mask = clean.str.fullmatch(four_number_year_pattern).fillna(False)
    partial_month_mask = clean.str.fullmatch(month_year_text_pattern).fillna(False)
    year_only_mask = clean.str.fullmatch(year_only_pattern).fillna(False)
    ambiguous_two_digit_mask = clean.str.fullmatch(two_number_year_pattern).fillna(False)

    quality = pd.Series("unparseable", index=date_series.index, dtype="string")
    quality[missing_mask] = "missing"
    quality[full_date_month_mask | full_date_numeric_mask] = "full_date_reliable"
    quality[partial_month_mask] = "partial_month_year"
    quality[year_only_mask] = "partial_year_only"
    quality[ambiguous_two_digit_mask] = "ambiguous_two_digit_year"

    year = pd.Series(pd.NA, index=date_series.index, dtype="Int64")

    month_day_year_extract = (
        r"^(?:January|February|March|April|May|June|July|August|September|October|November|December)"
        r"\s+(?:(?:1|21|31)st|(?:2|22)nd|(?:3|23)rd|(?:[4-9]|1[0-9]|20|2[4-9]|30)th)"
        r"(?:,\s*)?\s+(\d{4})$"
    )
    month_year_extract = r"^(?:January|February|March|April|May|June|July|August|September|October|November|December)\s+(\d{4})$"
    four_digit_extract = r"^(?:0?[1-9]|1[0-2])/(?:0?[1-9]|[12][0-9]|3[01])/(\d{4})$"
    year_only_extract = r"^(\d{4})$"

    year.loc[full_date_month_mask] = clean.loc[full_date_month_mask].str.extract(month_day_year_extract, expand=False).astype("Int64")
    year.loc[full_date_numeric_mask] = clean.loc[full_date_numeric_mask].str.extract(four_digit_extract, expand=False).astype("Int64")
    year.loc[partial_month_mask] = clean.loc[partial_month_mask].str.extract(month_year_extract, expand=False).astype("Int64")
    year.loc[year_only_mask] = clean.loc[year_only_mask].str.extract(year_only_extract, expand=False).astype("Int64")
    year.loc[ambiguous_two_digit_mask] = pd.NA

    return year, quality

df_raw["publish_year"], df_raw["publish_year_quality"] = parse_year_and_flag(df_raw["publishDate"])
df_raw["first_publish_year"], df_raw["first_publish_year_quality"] = parse_year_and_flag(df_raw["firstPublishDate"])
```

### 10.6 ISBN Validity

This section will check whether ISBN can be used as an identifier, or does it contain placeholder or malformed values that could create false positives down the line during the project.

#### ISBN structure counts

```python
isbn = df_raw["isbn"].fillna("").astype(str).str.strip()

isbn_normalized = (
    isbn
    .str.upper()
    .str.replace("-", "", regex=False)
    .str.replace(" ", "", regex=False)
)

missing_like_values = {"", "NAN", "NONE", "NULL", "NA", "N/A"}
missing_mask = isbn_normalized.isin(missing_like_values)

length = isbn_normalized.str.len()

isbn_like_pattern = r"(?:\d{9}[\dX]|\d{13})"
isbn_like_mask = isbn_normalized.str.fullmatch(isbn_like_pattern).fillna(False)

placeholder_mask = (
    isbn_normalized.eq("9999999999999")
    | isbn_normalized.str.fullmatch(r"0+").fillna(False)
)

invalid_length_mask = (
    ~missing_mask
    & ~placeholder_mask
    & ~length.isin([10, 13])
)

invalid_characters_mask = (
    ~missing_mask
    & ~placeholder_mask
    & length.isin([10, 13])
    & ~isbn_like_mask
)

repeated_threshold = 10

repeat_base_mask = (
    ~missing_mask
    & ~placeholder_mask
    & isbn_like_mask
)

isbn_counts = isbn_normalized.where(repeat_base_mask).value_counts()

repeated_count = isbn_normalized.map(isbn_counts)

repeated_suspicious_mask = (
    repeat_base_mask
    & repeated_count.ge(repeated_threshold).fillna(False)
)

print("ISBN-like structure:", isbn_like_mask.sum())
print("Missing-like:", missing_mask.sum())
print("Placeholder-like:", placeholder_mask.sum())
print("Invalid length:", invalid_length_mask.sum())
print("Invalid characters with valid length:", invalid_characters_mask.sum())
print("Repeated suspicious:", repeated_suspicious_mask.sum())
```

| Missing-like |   0 |
| Placeholder-like             |   4354 |
| Invalid length                     |   28 |
| Invalid characters with valid length  |   4699 |

Top repeated normalized ISBN value is 9999999999999, this seems to be used as a placeholder value for famous or classical books. No other ISBN is repeated as much as 9999999999999. Repeated ISBN-like values consistently match the same title and author, but a small number have edition-level metadata differences. ISBN is useful as a duplicate signal after excluding placeholders and invalid identifiers, but it should not be the only automatic deletion rule.

The rows/books with ISBN 9999999999999 should stay in the dataset, but placeholder ISBN value should be excluded from ISBN-based duplicate matching. During cleaning create a normalized column and a flag column to show the quality of the ISBN and excluding placeholders and invalid structures.

#### ISBN summary table

```python
valid_for_isbn_duplicate_check = (
    ~missing_mask
    & ~placeholder_mask
    & isbn_like_mask
)

isbn_duplicate_base = df_raw.loc[
    valid_for_isbn_duplicate_check,
    ["title", "author", "isbn", "bookFormat", "publishDate", "publisher", "language"]
].copy()

isbn_duplicate_base["isbn_normalized"] = isbn_normalized[valid_for_isbn_duplicate_check]

isbn_group_summary = (
    isbn_duplicate_base
    .groupby("isbn_normalized")
    .agg(
        rows=("isbn_normalized", "size"),
        unique_titles=("title", lambda x: x.nunique(dropna=False)),
        unique_authors=("author", lambda x: x.nunique(dropna=False)),
        unique_formats=("bookFormat", lambda x: x.nunique(dropna=False)),
        unique_publish_dates=("publishDate", lambda x: x.nunique(dropna=False)),
        unique_publishers=("publisher", lambda x: x.nunique(dropna=False)),
        unique_languages=("language", lambda x: x.nunique(dropna=False)),
    )
    .reset_index()
)

duplicated_isbn_groups = isbn_group_summary.query("rows > 1")
```

```python
isbn_duplicate_metrics = pd.Series({
    "duplicated non-placeholder ISBN groups": duplicated_isbn_groups.shape[0],

    "repeated ISBN rows": duplicated_isbn_groups["rows"].sum(),

    "groups with same title + author": (
        (duplicated_isbn_groups["unique_titles"] == 1)
        & (duplicated_isbn_groups["unique_authors"] == 1)
    ).sum(),

    "groups with differing format/date/publisher/language": (
        (duplicated_isbn_groups["unique_formats"] > 1)
        | (duplicated_isbn_groups["unique_publish_dates"] > 1)
        | (duplicated_isbn_groups["unique_publishers"] > 1)
        | (duplicated_isbn_groups["unique_languages"] > 1)
    ).sum(),

    "groups with differing title or author": (
        (duplicated_isbn_groups["unique_titles"] > 1)
        | (duplicated_isbn_groups["unique_authors"] > 1)
    ).sum(),
})

isbn_duplicate_metrics
```
