#### 3. Schema Inspection

```python
df_raw.info()
```

The dataset contains 52478 rows and 25 columns. 20 columns are categorical and 5 are numeric. Genres, characters, awards, ratingsbystarts and setting are high-cardinality columns. Bookid is the ID column of the dataset. Some columns contain lists but have a Dtype of str.

```python
schema_audit = pd.DataFrame({
    "column": df_raw.columns,
    "dtype": df_raw.dtypes.astype(str).values,
    "non_null_count": df_raw.notna().sum().values,
    "missing_count": df_raw.isna().sum().values,
    "missing_percent": (df_raw.isna().mean() * 100).round(2).values,
    "unique_values": [df_raw[col].nunique(dropna=True) for col in df_raw.columns]
})

schema_audit.sort_values("missing_percent", ascending=False)
```

#### 4. Missingness audit

```python
missing = df_raw.isnull().mean().sort_values(ascending=False).mul(100).round(2)
missing
```

The highest missingness appears in `edition`, `series`, `firstPublishDate`, and `price`.

These missing values should not be handled the same way:
- `edition` is likely optional metadata and may not be useful for core analysis.
- `series` missingness may be meaningful because many books are standalone.
- `firstPublishDate` is important for temporal analysis but has substantial missingness.
- `price` needs further investigation before deciding whether to use it.

#### 5. Duplicate Audit

```python
exact_duplicate_count = df_raw.duplicated().sum()
exact_duplicate_count
```

```python
exact_duplicates = df_raw[df_raw.duplicated(keep=False)].sort_values("bookId")
exact_duplicates.head(10)
```

```python
bookid_duplicate_count = df_raw.duplicated(subset=["bookId"]).sum()
bookid_duplicate_count
```

```python
title_author_duplicates = df_raw[
    df_raw.duplicated(subset=["title", "author"], keep=False)
].sort_values(["title", "author"])

title_author_duplicates[["bookId", "title", "author", "isbn", "rating", "numRatings"]].head(20)
```

Exact duplicate rows can likely be removed during cleaning because they repeat the same record.

However, `title + author` duplicates require caution. These may represent different editions, formats, translations, or duplicate scraped records. They should not be automatically dropped without further investigation.
